```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().mean()
```

```python
x = df.drop(columns=['Survived'])

y=df['Survived']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
x_train.shape,y_train.shape
```

```python
x_train.isnull().mean()
```

```python
x_test.isnull().mean()
```

```python
mean_age = x_train['Age'].mean()
median_age = x_train['Age'].median()

mean_fare = x_train['Fare'].mean()
median_fare = x_train['Fare'].median()
```

```python
mean_age,median_age,mean_fare,median_fare
```

```python
x_train['Age_mean'] = x_train['Age'].fillna(mean_age)
x_train['Age_median'] = x_train['Age'].fillna(median_age)
x_train['Fare_mean'] = x_train['Fare'].fillna(mean_fare)
x_train['Fare_median'] = x_train['Fare'].fillna(median_fare)
```

```python
x_train.sample(5)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
x_train['Age'].plot(kind='kde',ax=ax,color = 'orange')
x_train['Age_median'].plot(kind='kde',ax=ax,color = 'red')
x_train['Age_mean'].plot(kind = 'kde',ax = ax,color = 'blue')

lines,labels = ax.get_legend_handles_labels()
ax.legend(lines,labels,loc = 'best')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
x_train['Fare'].plot(kind='kde',ax=ax,color = 'orange')
x_train['Fare_median'].plot(kind='kde',ax=ax,color = 'red')
x_train['Fare_mean'].plot(kind = 'kde',ax = ax,color = 'blue')

lines,labels = ax.get_legend_handles_labels()
ax.legend(lines,labels,loc = 'best')
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=2
)
```

```python
trf = ColumnTransformer([
    ('imputer1',SimpleImputer(strategy='mean'),['Age']),
    ('imputer2',SimpleImputer(strategy='median'),['Fare']),
],remainder = 'passthrough')
```

```python
trf.fit(x_train)
```

```python
x_train = trf.transform(x_train)
x_test = trf.transform(x_test)
```
